```python
import pandas as pd
import numpy as npm
```

```python
df = pd.read_csv('kaggle_data2.csv')

print(df.head())
```

```python
# filter by LOCATION with Name sunsari, but it is in back with comma, i only want to filter with sunsari, so i will use str.contains() method to filter the data
df_sunsari = df[df['LOCATION'].str.contains('sunsari', case=False, na=False)]
print(df_sunsari.head())

```

```python

df_sunsari.to_csv('data2.csv', index=False)
```
